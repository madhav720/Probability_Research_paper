```python

# ================================================================
# DATASET 3: ROAD NOISE RISK LEVEL PREDICTION
# COMPLETE JUPYTER NOTEBOOK CODE
# ================================================================
# Models: LR, LMR, MLR, RF, XGB
# Splits: 80/20, 70/30, 60/40
#
# Inline graphs:
# 1. Accuracy comparison
# 2. Precision, Recall and F1 comparison
# 3. Precision vs Recall — separate illustrative curves
# 4. Multiclass ROC curves
# 5. Confusion matrices
#
# Tables:
# 1. Dataset summary
# 2. Model performance
# 3. ROC-AUC scores
# 4. Robustness results
#
# Removed:
# - ROC-AUC comparison bar graph
# - Accuracy vs Perturbation graph
# - F1 vs Perturbation graph
#
# Graphs are displayed inline; no separate graph files are saved.
# ================================================================

%matplotlib inline

import warnings
warnings.filterwarnings("ignore")

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler, label_binarize
from sklearn.pipeline import make_pipeline
from sklearn.linear_model import LogisticRegression, LinearRegression
from sklearn.ensemble import RandomForestClassifier

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix,
    roc_curve,
    roc_auc_score,
    auc,
    precision_recall_curve
)

from IPython.display import display
from xgboost import XGBClassifier


# ================================================================
# 1. CONFIGURATION
# ================================================================

INPUT_FILE = "Road_noise_measurements_filled (2).xlsx"
SHEET_NAME = "Road_noise_measurements"

FEATURES = [
    "AVG_NOISE_LVL",
    "MEU",
    "Duration_hours",
    "Cumulative_MEU"
]

TARGET = "Risk_Level"

CLASS_NAMES = ["Low", "Moderate", "High"]
CLASS_CODES = {"Low": 0, "Moderate": 1, "High": 2}

MODEL_ORDER = ["LR", "LMR", "MLR", "RF", "XGB"]

SPLITS = [
    ("80/20", 0.20),
    ("70/30", 0.30),
    ("60/40", 0.40)
]

PERTURBATIONS = [20, 30, 40]
RANDOM_STATE = 42


# ================================================================
# 2. LOAD DATA
# ================================================================

df = pd.read_excel(INPUT_FILE, sheet_name=SHEET_NAME)

print("=" * 75)
print("ROAD NOISE RISK LEVEL PREDICTION")
print("=" * 75)
print("Original dataset shape:", df.shape)

required_columns = FEATURES + [TARGET]
missing_columns = [
    col for col in required_columns if col not in df.columns
]

if missing_columns:
    raise ValueError(f"Missing columns in dataset: {missing_columns}")

data = df[required_columns].copy()

for col in FEATURES:
    data[col] = pd.to_numeric(data[col], errors="coerce")

data[TARGET] = data[TARGET].astype("string").str.strip()
data = data.dropna(subset=required_columns)
data = data[data[TARGET].isin(CLASS_NAMES)].copy()

if data.empty:
    raise ValueError(
        "No usable rows remain. Check the file and Risk_Level labels."
    )

X = data[FEATURES].copy()
y = data[TARGET].map(CLASS_CODES).astype(int)

print("Rows after cleaning:", len(data))
print("Features:", FEATURES)
print("Target:", TARGET)


# ================================================================
# 3. DATASET SUMMARY
# ================================================================

print("\nCLASS DISTRIBUTION")

class_summary = (
    data[TARGET]
    .value_counts()
    .reindex(CLASS_NAMES, fill_value=0)
    .rename_axis("Risk Level")
    .to_frame("Samples")
)

class_summary["Percentage (%)"] = (
    class_summary["Samples"] / len(data) * 100
).round(2)

display(class_summary)

print("\nFEATURE STATISTICS")
display(X.describe().round(3))

print("\nMISSING VALUES AFTER CLEANING")
display(data[required_columns].isnull().sum().to_frame("Missing Values"))


# ================================================================
# 4. MODEL DEFINITIONS
# ================================================================

def create_models():
    return {
        # Multinomial logistic regression classifier
        "LR": make_pipeline(
            StandardScaler(),
            LogisticRegression(
                solver="lbfgs",
                max_iter=2000,
                class_weight="balanced",
                random_state=RANDOM_STATE
            )
        ),

        # Linear Regression baseline
        "LMR": make_pipeline(
            StandardScaler(),
            LinearRegression()
        ),

        # Multiple Linear Regression
        # NOTE: With the same features, this is mathematically
        # identical to LMR. Both entries will produce duplicate results.
        "MLR": make_pipeline(
            StandardScaler(),
            LinearRegression()
        ),

        "RF": RandomForestClassifier(
            n_estimators=200,
            random_state=RANDOM_STATE,
            class_weight="balanced",
            n_jobs=-1
        ),

        "XGB": XGBClassifier(
            objective="multi:softprob",
            num_class=3,
            n_estimators=150,
            max_depth=5,
            learning_rate=0.08,
            subsample=0.9,
            colsample_bytree=0.9,
            eval_metric="mlogloss",
            random_state=RANDOM_STATE,
            n_jobs=-1
        )
    }


# ================================================================
# 5. HELPER FUNCTIONS
# ================================================================

def predict_labels(model_name, model, X_test):
    """Predict class codes 0, 1 and 2."""

    predictions = np.asarray(model.predict(X_test)).reshape(-1)

    if model_name in ["LMR", "MLR"]:
        predictions = np.rint(predictions)
        predictions = np.clip(predictions, 0, 2)

    return predictions.astype(int)


def get_class_scores(model_name, model, X_test):
    """
    Return one score column for each class.

    Classifiers use predicted probabilities.
    Linear regression uses negative distance to each class code;
    these scores are not calibrated probabilities.
    """

    if model_name not in ["LMR", "MLR"]:
        raw_scores = model.predict_proba(X_test)

        final_model = (
            list(model.named_steps.values())[-1]
            if hasattr(model, "named_steps")
            else model
        )

        model_classes = getattr(
            final_model, "classes_", np.array([0, 1, 2])
        )

        scores = np.zeros((len(X_test), 3))

        for index, class_code in enumerate(model_classes):
            scores[:, int(class_code)] = raw_scores[:, index]

        return scores

    predictions = np.asarray(model.predict(X_test)).reshape(-1)

    return -np.column_stack([
        np.abs(predictions - class_code)
        for class_code in range(3)
    ])


def calculate_metrics(y_true, y_pred):
    return {
        "Accuracy (%)": 100 * accuracy_score(y_true, y_pred),
        "Precision (%)": 100 * precision_score(
            y_true, y_pred, average="weighted", zero_division=0
        ),
        "Recall (%)": 100 * recall_score(
            y_true, y_pred, average="weighted", zero_division=0
        ),
        "F1 (%)": 100 * f1_score(
            y_true, y_pred, average="weighted", zero_division=0
        )
    }


def weighted_ovr_auc(y_true, scores):
    """Support-weighted one-vs-rest ROC-AUC."""

    auc_values = []
    supports = []

    for class_code in range(3):
        binary_true = (np.asarray(y_true) == class_code).astype(int)

        if len(np.unique(binary_true)) < 2:
            continue

        auc_value = roc_auc_score(
            binary_true, scores[:, class_code]
        )

        auc_values.append(auc_value)
        supports.append(int(binary_true.sum()))

    if not supports:
        return np.nan

    return float(np.average(auc_values, weights=supports))


# ================================================================
# 6. TRAINING AND EVALUATION
# ================================================================

performance_rows = []
auc_rows = []
robustness_rows = []

roc_data = {}
confusion_data = {}
trained_models = {}

for split_name, test_size in SPLITS:

    print("\n" + "=" * 75)
    print(f"TRAIN/TEST SPLIT: {split_name}")
    print("=" * 75)

    X_train, X_test, y_train, y_test = train_test_split(
        X,
        y,
        test_size=test_size,
        random_state=RANDOM_STATE,
        stratify=y
    )

    trained_models[split_name] = {}

    for model_name, model in create_models().items():

        print(f"Training {model_name}...")

        model.fit(X_train, y_train)

        y_pred = predict_labels(model_name, model, X_test)
        scores = get_class_scores(model_name, model, X_test)

        metrics = calculate_metrics(y_test, y_pred)

        performance_rows.append({
            "Split": split_name,
            "Model": model_name,
            "Train Samples": len(X_train),
            "Test Samples": len(X_test),
            **metrics
        })

        trained_models[split_name][model_name] = model

        confusion_data[(split_name, model_name)] = confusion_matrix(
            y_test, y_pred, labels=[0, 1, 2]
        )

        # Per-class one-vs-rest ROC curves
        per_class_auc = []

        for class_code, class_name in enumerate(CLASS_NAMES):

            binary_true = (
                y_test.to_numpy() == class_code
            ).astype(int)

            if len(np.unique(binary_true)) < 2:
                continue

            fpr, tpr, _ = roc_curve(
                binary_true, scores[:, class_code]
            )

            class_auc = auc(fpr, tpr)
            per_class_auc.append(class_auc)

            roc_data[
                (split_name, model_name, class_name)
            ] = (fpr, tpr, class_auc)

        macro_auc = (
            float(np.mean(per_class_auc))
            if per_class_auc else np.nan
        )

        try:
            weighted_auc = weighted_ovr_auc(y_test, scores)
        except ValueError:
            weighted_auc = np.nan

        auc_rows.append({
            "Split": split_name,
            "Model": model_name,
            "Macro AUC (OvR)": macro_auc,
            "Weighted AUC (OvR)": weighted_auc
        })

        print(
            f"  Accuracy={metrics['Accuracy (%)']:.2f}% | "
            f"Precision={metrics['Precision (%)']:.2f}% | "
            f"Recall={metrics['Recall (%)']:.2f}% | "
            f"F1={metrics['F1 (%)']:.2f}%"
        )

        # --------------------------------------------------------
        # ROBUSTNESS ANALYSIS
        # Metrics are recorded in a table; no perturbation graphs.
        # --------------------------------------------------------

        feature_std = X_train.std().replace(0, 1).fillna(1)

        for perturbation in PERTURBATIONS:

            rng = np.random.default_rng(
                RANDOM_STATE + perturbation
            )

            noise = rng.normal(
                loc=0,
                scale=perturbation / 100,
                size=X_test.shape
            )

            X_noisy = pd.DataFrame(
                X_test.to_numpy()
                + noise * feature_std.to_numpy(),
                columns=FEATURES,
                index=X_test.index
            )

            noisy_pred = predict_labels(
                model_name, model, X_noisy
            )

            noisy_metrics = calculate_metrics(y_test, noisy_pred)

            robustness_rows.append({
                "Split": split_name,
                "Model": model_name,
                "Perturbation (%)": perturbation,
                **noisy_metrics
            })


# ================================================================
# 7. RESULTS TABLES
# ================================================================

results_df = pd.DataFrame(performance_rows)
auc_df = pd.DataFrame(auc_rows)
robustness_df = pd.DataFrame(robustness_rows)

print("\nMODEL PERFORMANCE RESULTS")
display(
    results_df
    .sort_values(["Split", "Model"])
    .round(3)
    .reset_index(drop=True)
)

print("\nROC-AUC SCORES")
display(
    auc_df
    .sort_values(["Split", "Model"])
    .round(4)
    .reset_index(drop=True)
)

print("\nROBUSTNESS RESULTS")
display(
    robustness_df
    .sort_values(["Split", "Model", "Perturbation (%)"])
    .round(3)
    .reset_index(drop=True)
)


# ================================================================
# 8. ACCURACY COMPARISON
# ================================================================

for split_name, _ in SPLITS:

    plot_data = (
        results_df[results_df["Split"] == split_name]
        .set_index("Model")
        .reindex(MODEL_ORDER)
    )

    fig, ax = plt.subplots(figsize=(11, 6))

    plot_data["Accuracy (%)"].plot(
        kind="bar", ax=ax, rot=0
    )

    ax.set_title(f"Accuracy Comparison - {split_name}")
    ax.set_xlabel("Model")
    ax.set_ylabel("Accuracy (%)")
    ax.set_ylim(0, 110)
    ax.grid(axis="y", alpha=0.3)

    for container in ax.containers:
        labels = [
            f"{bar.get_height():.2f}%"
            if np.isfinite(bar.get_height()) else ""
            for bar in container
        ]
        ax.bar_label(container, labels=labels, padding=3, fontsize=8)

    plt.tight_layout()
    plt.show()


# ================================================================
# 9. PRECISION, RECALL AND F1 COMPARISON
# ================================================================

for split_name, _ in SPLITS:

    plot_data = (
        results_df[results_df["Split"] == split_name]
        .set_index("Model")
        .reindex(MODEL_ORDER)
    )

    metrics_to_plot = [
        "Precision (%)",
        "Recall (%)",
        "F1 (%)"
    ]

    ax = plot_data[metrics_to_plot].plot(
        kind="bar",
        figsize=(14, 7),
        rot=0,
        width=0.8
    )

    ax.set_title(
        f"Precision, Recall and F1 Comparison - {split_name}"
    )
    ax.set_xlabel("Model")
    ax.set_ylabel("Score (%)")
    ax.set_ylim(0, 115)
    ax.grid(axis="y", alpha=0.3)

    for container in ax.containers:
        labels = [
            f"{bar.get_height():.2f}%"
            if np.isfinite(bar.get_height()) else ""
            for bar in container
        ]
        ax.bar_label(
            container,
            labels=labels,
            padding=3,
            fontsize=8,
            rotation=90
        )

    plt.tight_layout()
    plt.show()


# ================================================================
# 10. PRECISION VS RECALL — SKETCH-STYLE CURVES
# ================================================================
# IMPORTANT:
# Smooth curves below are illustrative visual guides.
# Markers represent the actual measured precision/recall values.
# These curves must not be described as measured PR curves.
# ================================================================

for split_name, _ in SPLITS:

    plot_data = (
        results_df[results_df["Split"] == split_name]
        .set_index("Model")
        .reindex(MODEL_ORDER)
        .reset_index()
    )

    valid_data = plot_data.dropna(
        subset=["Recall (%)", "Precision (%)"]
    )

    fig, ax = plt.subplots(figsize=(13, 8))

    # Separate curves: steep rise followed by flattening.
    # They end at each model's actual measured recall/precision.
    t = np.linspace(0, 1, 250)

    for _, row in valid_data.iterrows():

        model_name = row["Model"]
        recall = float(row["Recall (%)"])
        precision = float(row["Precision (%)"])

        # Illustrative curve only — not empirical intermediate points.
        curve_recall = recall * t
        curve_precision = precision * (
            (1 - np.exp(-7 * t)) / (1 - np.exp(-7))
        )

        ax.plot(
            curve_recall,
            curve_precision,
            linewidth=2,
            label=model_name
        )

        # Real measured result at the curve endpoint
        ax.scatter(
            recall,
            precision,
            s=75,
            zorder=5
        )

    # Model names are placed at the upper-right, outside the line area.
    # Use the actual endpoint values in the text.
    endpoint_order = (
        valid_data
        .sort_values("Recall (%)")
        .reset_index(drop=True)
    )

    for index, row in endpoint_order.iterrows():

        model_name = row["Model"]
        recall = float(row["Recall (%)"])
        precision = float(row["Precision (%)"])

        ax.annotate(
            f"{model_name}: P={precision:.2f}%, R={recall:.2f}%",
            xy=(recall, precision),
            xycoords="data",
            xytext=(8, 12 + index * 20),
            textcoords="offset points",
            fontsize=8,
            fontweight="bold",
            bbox=dict(
                boxstyle="round,pad=0.3",
                facecolor="white",
                edgecolor="gray",
                alpha=0.95
            ),
            arrowprops=dict(
                arrowstyle="-",
                color="gray",
                linewidth=0.8
            ),
            annotation_clip=False
        )

    ax.set_title(
        f"Precision vs Recall - {split_name}",
        fontsize=14
    )
    ax.set_xlabel("Recall (%)")
    ax.set_ylabel("Precision (%)")
    ax.set_xlim(0, 110)
    ax.set_ylim(0, 115)
    ax.grid(True, alpha=0.3)

    # Legend outside the plotting area
    ax.legend(
        title="Models",
        loc="lower right",
        fontsize=9
    )

    fig.subplots_adjust(
        left=0.10,
        right=0.78,
        top=0.90,
        bottom=0.12
    )

    plt.show()


# ================================================================
# 11. MULTICLASS ROC CURVES
# ================================================================

for split_name, _ in SPLITS:

    fig, ax = plt.subplots(figsize=(13, 8))

    for model_name in MODEL_ORDER:
        for class_name in CLASS_NAMES:

            key = (split_name, model_name, class_name)

            if key not in roc_data:
                continue

            fpr, tpr, class_auc = roc_data[key]

            ax.plot(
                fpr,
                tpr,
                linewidth=1.6,
                label=(
                    f"{model_name} - {class_name} "
                    f"(AUC={class_auc:.3f})"
                )
            )

    ax.plot(
        [0, 1], [0, 1],
        linestyle="--",
        color="black",
        label="Random classifier"
    )

    ax.set_title(f"Multiclass ROC Curves - {split_name}")
    ax.set_xlabel("False Positive Rate")
    ax.set_ylabel("True Positive Rate")
    ax.set_xlim(0, 1)
    ax.set_ylim(0, 1.02)
    ax.grid(True, alpha=0.3)
    ax.legend(loc="lower right", fontsize=8, ncol=2)

    plt.tight_layout()
    plt.show()


# ================================================================
# 12. CONFUSION MATRICES
# ================================================================

for split_name, _ in SPLITS:
    for model_name in MODEL_ORDER:

        cm = confusion_data[(split_name, model_name)]

        fig, ax = plt.subplots(figsize=(7, 6))

        image = ax.imshow(cm, interpolation="nearest")
        fig.colorbar(image, ax=ax)

        ax.set_title(
            f"Confusion Matrix - {model_name} ({split_name})"
        )
        ax.set_xticks(range(3))
        ax.set_yticks(range(3))
        ax.set_xticklabels(CLASS_NAMES)
        ax.set_yticklabels(CLASS_NAMES)
        ax.set_xlabel("Predicted Risk Level")
        ax.set_ylabel("Actual Risk Level")

        threshold = cm.max() / 2 if cm.size else 0

        for i in range(3):
            for j in range(3):
                ax.text(
                    j, i, str(cm[i, j]),
                    ha="center",
                    va="center",
                    color="white" if cm[i, j] > threshold else "black"
                )

        plt.tight_layout()
        plt.show()


# ================================================================
# 13. FINAL COMPARISON TABLE
# ================================================================

summary_df = results_df.merge(
    auc_df,
    on=["Split", "Model"],
    how="left"
)

print("\nFINAL MODEL COMPARISON WITH AUC")
display(
    summary_df
    .sort_values(["Split", "F1 (%)"], ascending=[True, False])
    .round(4)
    .reset_index(drop=True)
)

print("\nBEST MODEL BY WEIGHTED F1 FOR EACH SPLIT")

best_indices = summary_df.groupby("Split")["F1 (%)"].idxmax()

display(
    summary_df.loc[
        best_indices,
        [
            "Split",
            "Model",
            "Accuracy (%)",
            "Precision (%)",
            "Recall (%)",
            "F1 (%)",
            "Macro AUC (OvR)",
            "Weighted AUC (OvR)"
        ]
    ]
    .sort_values("Split")
    .round(4)
    .reset_index(drop=True)
)


# ================================================================
# 14. COMPLETION
# ================================================================

print("\n" + "=" * 75)
print("ROAD NOISE RISK LEVEL ANALYSIS COMPLETED")
print("=" * 75)
print("Models:", ", ".join(MODEL_ORDER))
print("Splits: 80/20, 70/30, 60/40")
print("All tables and graphs are displayed inline.")
print("No separate graph image files are saved.")
print("ROC-AUC comparison bar graph: removed.")
print("Accuracy vs Perturbation graph: removed.")
print("F1 vs Perturbation graph: removed.")
print("=" * 75)
```
